```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
nfs = pd.read_csv("nfs_multifile_results.csv")
fuse = pd.read_csv("fuse_multifile_results.csv")

nfs["FileNumber"] = range(1, len(nfs) + 1)
fuse["FileNumber"] = range(1, len(fuse) + 1)

plt.figure(figsize=(8, 5))
plt.plot(nfs["FileNumber"], nfs["CumulativeTime(ms)"], marker='o', label="NFS")
plt.plot(fuse["FileNumber"], fuse["CumulativeTime(ms)"], marker='o', label="FUSE")

plt.yscale("log") 
plt.xlabel("File Number (1–5)")
plt.ylabel("Cumulative Time (ms, log scale)")
plt.title("Cumulative Time Comparison")
plt.legend()
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)

plt.show()
```

```python
# Load results
nfs = pd.read_csv("nfs_multifile_results.csv")
fuse = pd.read_csv("fuse_multifile_results.csv")

# Extract file number (1 through N)
nfs["FileNumber"] = range(1, len(nfs) + 1)
fuse["FileNumber"] = range(1, len(fuse) + 1)

# Bar width and positions
x = np.arange(len(nfs))  # file numbers as positions
width = 0.35

plt.figure(figsize=(8, 5))

# Bars for NFS and FUSE
plt.bar(x - width/2, nfs["OperationTime(ms)"], width, label="NFS")
plt.bar(x + width/2, fuse["OperationTime(ms)"], width, label="FUSE")

# Labels & formatting
plt.yscale("log")
plt.xlabel("File Number (1–5)")
plt.ylabel("Operation Time (ms, log scale)")
plt.title("Individual Operation Time Comparison")
plt.xticks(x, nfs["FileNumber"])
plt.legend()
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)

plt.show()
```
